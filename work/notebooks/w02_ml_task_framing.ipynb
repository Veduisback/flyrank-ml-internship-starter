## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

### My answer

**ML task type: Ranking**

My provisional lane is Refresh / Content Opportunity Scoring. I frame this as a ranking problem because the decision is to order pages by priority for human review, rather than simply classify each page as refresh or no-refresh.

The model would use observable page-level signals to produce a prioritized review queue. A higher-ranked page would be reviewed earlier by the content or SEO team. The model is decision-support; the human reviewer still decides whether to refresh, expand, protect, prune, or monitor the page.


```python
print("ML task type: Ranking")
print("Decision: prioritize pages for human review")
print("Output: ranked review queue")
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

### My answer

**Primary metric: Precision@K**

I would use Precision@K because the content team has limited review capacity, so the quality of the top part of the ranked queue matters most.

For example, Precision@50 would measure how many of the 50 highest-priority pages match the defined target or proxy. The exact K should reflect a realistic review capacity.

I would compare the model against a simple fixed-rule or baseline ranking to check whether the additional complexity of ML provides useful improvement.


```python
print("Primary metric: Precision@K")
print("Example: Precision@50")
print("Reason: review capacity is limited")
```
