## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

### My answer

**ML task type: Ranking**

My provisional lane is Refresh / Content Opportunity Scoring. I frame this as a ranking problem because the decision is to order pages by priority for human review, rather than simply classify each page as refresh or no-refresh.

The model would use observable page-level signals to produce a prioritized review queue. A higher-ranked page would be reviewed earlier by the content or SEO team. The model is decision-support; the human reviewer still decides whether to refresh, expand, protect, prune, or monitor the page.


```python
print("ML task type: Ranking")
print("Decision: prioritize pages for human review")
print("Output: ranked review queue")
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

### My answer

For this starter exercise, I will use `trend_direction` as a **proxy target**. It is an observed current-window signal describing the page's recent direction, rather than a true future outcome.

For the eventual capstone, I would prefer a leakage-safe target based on an earlier feature window and a later measured outcome window. That would make the prediction task closer to the real decision of identifying pages that may need attention in the future.


```python
print("Proxy target: trend_direction")
print("Proxy type: observed current-window directional signal")
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

### My answer

**Primary metric: Precision@K**

I would use Precision@K because the content team has limited review capacity, so the quality of the top part of the ranked queue matters most.

For example, Precision@50 would measure how many of the 50 highest-priority pages match the defined target or proxy. The exact K should reflect a realistic review capacity.

I would compare the model against a simple fixed-rule or baseline ranking to check whether the additional complexity of ML provides useful improvement.


```python
print("Primary metric: Precision@K")
print("Example: Precision@50")
print("Reason: review capacity is limited")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## 5. Why ML beats a fixed rule here

A fixed rule could use one or two conditions, such as sending pages with declining trends or older content to the review queue. The difficulty is that page priority can depend on several signals at the same time.

ML can combine multiple measured signals and produce a consistent ranking across many pages instead of requiring manually chosen thresholds for every signal.

However, ML should only be used if it improves the decision compared with a transparent baseline. The baseline should remain part of the evaluation so that extra model complexity has to demonstrate useful improvement.
